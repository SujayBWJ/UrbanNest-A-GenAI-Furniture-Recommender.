# RAG Testing Notebook
Notebook used for testing retrieval quality and conversational recommendations.


```
import pandas as pd
products = pd.read_csv('../rag/data/furniture_products.csv')
products.head()
```

## Initial Observation
Some duplicate product names were present in the dataset. Basic cleaning was done before embedding generation.


```
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
sample = products['product_name'].tolist()[:10]
embeddings = model.encode(sample)
print('Embedding shape:', embeddings.shape)
```
