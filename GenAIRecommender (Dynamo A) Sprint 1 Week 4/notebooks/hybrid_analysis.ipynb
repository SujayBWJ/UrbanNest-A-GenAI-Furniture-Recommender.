# Hybrid Recommender Architecture

The hybrid recommender combines multiple recommendation signals to improve personalization and recommendation quality.

## Signals Used
1. Collaborative Filtering Score
2. Embedding Similarity Score
3. Price Match Score

## Workflow
1. Generate recommendations using collaborative filtering
2. Generate content-based recommendations using embeddings
3. Combine recommendation candidates
4. Apply weighted scoring
5. Rank items and return Top-K recommendations

## Benefits
- Better personalization
- Improved diversity
- Reduced cold-start issues
- More relevant recommendations



# Signal Weighting Strategy

The hybrid model combines multiple recommendation signals using weighted aggregation.

## Weight Distribution
- Collaborative Filtering: 40%
- Embedding Similarity: 40%
- Price Match: 20%

## Formula

Final Score =
0.4 × CF Score +
0.4 × Embedding Score +
0.2 × Price Match Score

## Reasoning
- Collaborative filtering captures behavioral similarity
- Embeddings capture semantic similarity
- Price matching improves affordability relevance

# Evaluation Comparison

## Models Compared
1. Popularity-Based Recommender
2. Collaborative Filtering
3. Hybrid Recommender

## Metrics
- Precision@K
- NDCG@K

## Observations
- Popularity model gives generic recommendations
- Collaborative filtering improves personalization
- Hybrid recommender provides the best balance between relevance and diversity

## Conclusion
The hybrid model outperformed baseline approaches by combining multiple recommendation signals.
